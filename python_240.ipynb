```python
#Excercise 7: Write a program that will take three digits from the user and add the square of eachdigit.
num = int(input("enter number: "))
n = len(str(num))
sum = 0
for i in range(0,n):
  a = num % 10
  num = num // 10
  sum += a **2

print(sum)
```

```python
#Excercise 8: Write a program that will check whether the number is armstrong number or not.
# if a number is armstrong  then
#An Armstrong number is one whose sum of digits raised to the power three equals the number itself. 371,
#for example, is an Armstrong number because 3**3 + 7**3 + 1**3 = 371.1634 = 1** 4 + 6**4 + 3**4 + 4**4
num = int(input("enter number: "))
num1 = num
n = len(str(num))
num2 = 0
for i in range(0,n):
  a = num % 10
  num = num // 10
  num2 += a ** n

print(num2)

print(f"{num1} is armstrong number") if (num1 == num2) else print(f"{num1} is not an armstrong number")
```
