```python
class Vehicle:
  """docstring"""
  def __int__(self,color,doors,tires):
    """constructor"""
    self.color = color
    self.doors = doors
    self.tires = tires

  def brake(self):
    """stop the car"""
    return "Breaks applied"

  def drive(self):
    """driving the car"""
    return "car is moving"
```

```python
class Car(Vehicle):
  """the car sub class"""
  def brake(self):
    """override brake method"""
    return "override breaks"

if __name__=="__main__":
  car = Car("yellow",2,4)
  car.brake()
  car.drive()
```

```python
def my_decorator(func):
  def wrapper(*args, **kwargs):
    print("Something is happening before the function is called.")
    result = func(*args, **kwargs)
    print("Something is happening after the function is called.")
    return result
  return wrapper

@my_decorator
def say_hello():
  print("Hello!")

say_hello()
```
