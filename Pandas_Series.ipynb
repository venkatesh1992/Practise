### .loc() and .iloc()

*  **.loc** is label based extraction
*  **.iloc** is index based extraction


```python
label_alphas['label_I':'label_M']
```

```python
# loc is faster when compared to array notation
label_alphas.loc['label_I':'label_M']
```
