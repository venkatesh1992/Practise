## different ways of creating dataframe in pandas


1.   by using read files **pd.read_csv('filepath')**
2.   by using read_excel **pd.read_excel('filepath',sheetname)**
3.   by using **pd.DataFrame(python dictionary)**
4.   by using **pd.DataFrame(list of tuples,column names)**
5.   by using **pd.DataFrame(list of dictionaries)**

```python
#by using read files pd.read_csv('filepath')
df_csv=pd.read_csv('weather_data.csv')
```

```python
#3.   by using **pd.DataFrame(python dictionary)**
weather_data={
    'day': ['1/1/2017','1/2/2017','1/3/2017'],
    'temperature':[32,35,28],
    'windspeed':[6,7,2],
    'event':['Rain','Sunny','Snow']
}
df_dict = pd.DataFrame(weather_data)
df_dict
```

```python
#now we will see reading excel file
df = pd.read_excel('stock_data.xlsx','Sheet1')
df
```

```python
#while reading excel file we have additional methods like converters to clean the messsy data
def convert_people_cell(cell):
  if cell == 'n.a.':
    return 'sam walton'
  return cell

# converters will take method in dictionary
df = pd.read_excel('stock_data.xlsx','Sheet1',converters={
    'people':convert_people_cell
})
df
```

```python
#similarly if i want to convert other cells,we can add additional methods
def convert_people_cell(cell):
  if cell == 'n.a.':
    return 'sam walton'
  return cell

def convert_eps_cell(cell):
  if cell == 'not available':
    return None
  return cell

# converters will take method in dictionary
df = pd.read_excel('stock_data.xlsx','Sheet1',converters={
    'people':convert_people_cell,
    'eps':convert_eps_cell
})
df
```

```python
#now we will see writing
df.to_excel("new.xlsx",sheet_name='stocks')
```

```python
#in excel , writing we can mention start row and start column(so that it will write from that point)
df.to_excel("new.xlsx",sheet_name='stocks',index=False,startrow=2,startcol=2)
```

**Handling missing Data**


1.   fillna
2.   interpolate
3.   dropna



```python
import pandas as pd
df = pd.read_csv('weather_data.csv')
df
```

```python
#i want to covert my date columns to date then
df = pd.read_csv('weather_data.csv',parse_dates=['day'])
df
```

```python
df
```

```python
# now i want to make day as index
df.set_index('day',inplace=True)
df
```

```python
# i want fill na with 0
df.fillna(0)
```

```python
#it is filling event also to 0, it looks weird to overcome that we will fill column wise
df.fillna({
    'temperature':0,
    'windspeed':0,
    'event':'No forecast'
})
```

```python
#instaed of making it to 0, we will fill that with previuos or after not null value, for better estimates
#for mean and average calculations
df.fillna(method='ffill') # previous data not null value
```

```python
df.fillna(method='bfill') # next day value
```

```python
# by using method='ffill'/bfill it is copying data vertically
# if i want my data to be copied horizontally then use axis and mention columns
df.fillna(method='bfill',axis='columns')
```

```python
df.fillna(method='ffill',axis='columns')
```

```python
#suppose i want to carry forward my value to next record only then we will use limit
df
df.fillna(method='ffill',limit=1)
```

```python
df.fillna(method='ffill',limit=2)
```

```python
#still the tempearture is not better estimate, to do that we have interpolate method in pandas
df.interpolate()
```

```python
#by default interpolate will do linaer guess, we can pass time as well
df.interpolate(method='time')
```

```python
# if i want to drop null values we have dropna method
df
```

```python
df.dropna()
# by default it will drop the full record, it it is having atleast one null value
```

```python
#if i want delete only all the values are null
df.dropna(how='all')
```

```python
# suppose i want to delete based on thresold of null values
# assume i want to drop record, if it has atleast two null values then we need to mention thresh=2
df.dropna(thresh=1) # it will drop it all the cols are null
```

```python
df.dropna(thresh=2)
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv('weather_data.csv')
df
```

```python
df.replace(-99999,np.NaN)
```

```python
#what if my data has mutiple values to replcae like -9999 and -8888, how to provide multiple values
# we can provide in list
df = pd.read_csv('weather_data.csv')
df
```

```python
df.replace([-99999,-8888],np.NAN)
```

```python
#still event has 0 value which is not correct,
#if i pass Zero in list and it will replace 0 in all the int columns which is not expected
df.replace([-99999,-8888,'0'],np.NAN)
```

```python
#to overcome above we can pass columns and replace values in dictionary
df.replace({
    'temperature':[-99999,-8888],
    'windspeed':[-99999,-8888],
    'event':'0'
},np.NaN)
```

```python
#what if my data has unit of measures also like 32F,12C,12mph like
# to replace those we need to use regex
df = pd.read_csv('weather_data.csv')
df
```

```python
df.replace('[a-zA-Z]','') # it didn't repalce, we need to mention regex=True
```

```python
df.replace('[a-zA-Z]','',regex=True)
#the problem is it replaces values in event also, which is not expected
```

```python
#to overcome we will pass regex with columns
df.replace({
    'temperature':'[a-zA-Z]',
    'windspeed':'[a-zA-Z]'
},'',regex=True)
```

```python
df1=pd.DataFrame({
    'city':['mumbai','delhi','bangalore'],
    'temperature':[35,40,30]})
df1
```

```python
#what if i have additional values/columns
df1=pd.DataFrame({
    'city':['mumbai','delhi','bangalore','hyderabad'],
    'temperature':[35,40,30,38]})
df1
```

```python
df2=pd.DataFrame({
    'city':['mumbai','bangalore','delhi','pune','kolkata'],
    'windspeed':[12,20,30,24,43]})
df2
```

```python
df3 = pd.merge(df1,df2,on='city')
df3
# by default it will do inner join
```

```python
df3 = pd.merge(df1,df2,on='city',how='inner')
df3
```

```python
df3 = pd.merge(df1,df2,on='city',how='left')
df3
```

```python
df3 = pd.merge(df1,df2,on='city',how='right')
df3
```

```python
df3 = pd.merge(df1,df2,on='city',how='outer')
df3
```

```python
#what if my dataframe has common columns
df1=pd.DataFrame({
    'city':['mumbai','delhi','bangalore'],
    'temperature':[35,40,30],
    'humidity':[60,80,20]
})
df1
```

```python
df2=pd.DataFrame({
    'city':['newyork','chicago','dallas'],
    'temperature':[12,13,14],
    'humidity':[20,30,40]
})
df2
```

```python
df3 = pd.merge(df1,df2,on='city')
df3
```

```python
df3 = pd.merge(df1,df2,on='city',how='outer')
df3
```

```python
#And if i want to know from which df my valu is derived we have option indicator
df3 = pd.merge(df1,df2,on='city',how='outer',indicator=True)
df3
```

```python
#Anf if i want my own suffixes instead of x,y i can mention in suffixes
df3 = pd.merge(df1,df2,on='city',how='outer',indicator=True,suffixes=['_left','_right'])
df3
```
